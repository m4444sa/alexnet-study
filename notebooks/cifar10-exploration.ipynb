# CIFAR-10 Dataset Exploration

This notebook explores CIFAR-10 before training an AlexNet-style convolutional neural network.

The project has two goals:

1. Reproduce and investigate important ideas from the 2012 AlexNet paper.
2. Study whether AlexNet learns an artificial shortcut when the training data contains a biased visual feature.

Before constructing the model, this notebook examines the dataset size, image shape, classes, and class distribution.


```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import torch
import torchvision
from torchvision import datasets, transforms

print("PyTorch version:", torch.__version__)
print("Torchvision version:", torchvision.__version__)
print("CUDA available:", torch.cuda.is_available())
```

```python
current_directory = Path.cwd()

if current_directory.name == "notebooks":
    project_directory = current_directory.parent
else:
    project_directory = current_directory

data_directory = project_directory / "data"

print("Project directory:", project_directory)
print("Data directory:", data_directory)
```

```python
basic_transform = transforms.ToTensor()

train_dataset = datasets.CIFAR10(
    root=data_directory,
    train=True,
    download=True,
    transform=basic_transform
)

test_dataset = datasets.CIFAR10(
    root=data_directory,
    train=False,
    download=True,
    transform=basic_transform
)
```

```python
print("Training examples:", len(train_dataset))
print("Test examples:", len(test_dataset))
print("Classes:", train_dataset.classes)
```

```python
#inspect an example
image, label = train_dataset[0]

print("Image shape:", image.shape)
print("Label number:", label)
print("Class name:", train_dataset.classes[label])
print("Minimum pixel value:", image.min().item())
print("Maximum pixel value:", image.max().item())
```

```python
#chack class distribution and plot it
training_labels = np.array(train_dataset.targets)
class_counts = np.bincount(training_labels)

for class_name, count in zip(train_dataset.classes, class_counts):
    print(f"{class_name}: {count}")
plt.figure(figsize=(10, 5))
plt.bar(train_dataset.classes, class_counts)
plt.title("CIFAR-10 Training Class Distribution")
plt.xlabel("Class")
plt.ylabel("Number of images")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show() 
```

```python
#display some images
figure, axes = plt.subplots(2, 5, figsize=(12, 5))

for class_index, axis in enumerate(axes.flat):
    example_index = training_labels.tolist().index(class_index)
    image, label = train_dataset[example_index]

    image = image.permute(1, 2, 0)

    axis.imshow(image)
    axis.set_title(train_dataset.classes[label])
    axis.axis("off")

plt.suptitle("One Training Example from Each CIFAR-10 Class")
plt.tight_layout()
plt.show()
```

## Initial observations

CIFAR-10 contains 50,000 training images and 10,000 test images divided evenly among ten classes. Each image has three colour channels and a resolution of 32 × 32 pixels.

The balanced class counts give us a controlled starting point. In the later ethics experiment, an artificial correlation between colour patches and labels will be introduced deliberately. This will allow us to distinguish class imbalance from shortcut bias.
